# Exercise 6: Structured vs. Unstructured Data

## What Is Structured Data?

Structured data is organized into a predefined format,
usually rows and columns, making it easy to store,
search, filter, and analyze.

## What Is Unstructured Data?

Unstructured data does not follow a predefined tabular
format. It often requires additional processing before
it can be analyzed.

## Classification of Data Examples

| Data Example | Classification | Explanation |
|---|---|---|
| Excel spreadsheet containing financial reports | Structured | Financial values are organized into rows and columns |
| Collection of photographs | Unstructured | Images contain visual information rather than predefined tabular fields |
| News articles | Unstructured | Articles contain free-form written text |
| Relational database of inventory records | Structured | Inventory information follows a defined database schema |
| Recorded customer interviews | Unstructured | Audio recordings contain spoken information that requires transcription or processing |

## Conclusion

Structured data is generally easier to analyze directly
using tools such as SQL and Pandas.

Unstructured data often requires preprocessing, such as
text extraction, transcription, or image recognition,
before meaningful analysis can be performed.



# Exercise 7: Transforming Unstructured Data into Structured Data

## Introduction

Unstructured data can be transformed into structured
data by extracting important information and organizing
it into predefined fields, such as rows and columns.

## 1. Travel Blog Posts

**Original format:** Free-form written text.

**Transformation process:**
- Use Natural Language Processing (NLP) to analyze text.
- Extract destinations, dates, hotels, and activities.
- Organize the extracted information into a table.

**Example structured fields:**
Destination, Travel Date, Hotel, Activities, Rating.

## 2. Customer Service Audio Recordings

**Original format:** Recorded conversations.

**Transformation process:**
- Convert speech into text using speech recognition.
- Identify customer complaints, requests, and sentiment.
- Extract information into structured categories.

**Example structured fields:**
Customer ID, Date, Issue Type, Sentiment, Resolution.

## 3. Handwritten Notes

**Original format:** Handwritten documents or images.

**Transformation process:**
- Scan or photograph the notes.
- Use Optical Character Recognition (OCR).
- Correct recognition errors and extract relevant details.
- Store the information in a spreadsheet or database.

**Example structured fields:**
Date, Topic, Description, Action Required.

## 4. Cooking Videos

**Original format:** Video recordings.

**Transformation process:**
- Transcribe spoken instructions.
- Identify ingredients and quantities.
- Extract cooking steps and preparation times.
- Organize the information into a recipe database.

**Example structured fields:**
Recipe Name, Ingredients, Quantities, Cooking Time, Steps.

## Conclusion

Transforming unstructured data into structured data
makes it easier to search, compare, visualize, and
analyze information using tools such as Pandas and SQL.
