## 1. What Is Machine Learning?

**Machine learning (ML)** is a field of artificial intelligence in which computers learn patterns from data to make predictions, classify information, identify groups, or improve decisions without a separate hand-written rule for every case.

For example, rather than manually defining every rule for detecting a suspicious transaction, an ML model can learn patterns from historical transactions and estimate whether a new transaction is unusual.

### Why is ML important for data analysts?

- **Prediction:** Estimate future demand, customer churn, or equipment failure.
- **Pattern discovery:** Detect relationships and segments that may not be obvious in summary tables.
- **Automation:** Analyze large volumes of records consistently and efficiently.
- **Decision support:** Translate historical data into measurable forecasts and risk estimates.

ML complements traditional data analysis: descriptive analytics explains *what happened*, while predictive ML can help estimate *what might happen*. Predictions still require careful validation, interpretation, and monitoring.

## 2. Applications Across Industries

| Industry | Specific application | Business value |
|---|---|---|
| **Finance** | Predict the probability that a borrower will default using historical repayment and application data | Improve risk assessment and identify applications requiring additional review |
| **Healthcare** | Analyze medical images to flag potentially abnormal findings for clinician review | Support faster prioritization and clinical workflows |
| **Retail / E-commerce** | Predict customer demand and recommend relevant products | Improve inventory planning and personalize recommendations |
| **Manufacturing** | Predict equipment failures from sensor measurements | Reduce unexpected downtime and plan maintenance |

**Important:** Models support human decisions; they can be inaccurate or biased, particularly when training data is incomplete or unrepresentative.

## 3. Three Main Types of Machine Learning

| Type | Definition | Example | Typical output |
|---|---|---|---|
| **Supervised learning** | Learns from examples containing input features **and known target labels/values**. | Predict whether a customer will cancel a subscription using past customer records labeled as churned/not churned. | Class or numeric prediction |
| **Unsupervised learning** | Discovers patterns or structures in data **without a predefined target label**. | Group retail customers by purchasing behavior for market segmentation. | Clusters or other hidden structure |
| **Reinforcement learning** | An agent learns a strategy through interactions with an environment, receiving rewards or penalties for actions. | Train a simulated robot to navigate a maze while rewarding progress and penalizing collisions. | Policy for choosing actions |

### Key distinctions

- **Supervised:** Has correct answers in training data; examples include classification and regression.
- **Unsupervised:** Has no labeled answer column; examples include clustering and dimensionality reduction.
- **Reinforcement:** Learns through sequential decisions and reward feedback rather than a static labeled dataset.

The appropriate type depends on the available data and the question being asked.

## 4. Developing a Machine Learning Model

### Stage 1 — Feature Selection

Choose the input variables that are useful, available at prediction time, and appropriate for the task. For loan default prediction, potential features include repayment history, debt-to-income ratio, and loan amount. Exclude identifiers, irrelevant fields, and **data leakage** (information that would only be known after the outcome).

**Why it matters:** Relevant features improve interpretability and may improve generalization. Poor features can add noise or unfair bias.

### Stage 2 — Model Selection

Choose an algorithm that matches the problem and available data. For a labeled yes/no outcome, logistic regression provides an interpretable baseline, while a random forest can capture nonlinear relationships. For unlabeled customer segmentation, K-means may be appropriate.

**Why it matters:** Different models have different trade-offs in accuracy, complexity, explainability, speed, and data requirements.

### Stage 3 — Model Evaluation

Test how well the trained model performs on **unseen data**. Split data into training and test sets (and use cross-validation where appropriate). For classification, assess precision, recall, F1-score, and confusion matrix—not accuracy alone when classes are imbalanced. For regression, consider MAE and RMSE. For clustering, consider silhouette score and practical interpretability.

**Why it matters:** Evaluation reveals whether a model generalizes, where it makes mistakes, and whether it is suitable for its intended use.

**Note:** In practice, defining the problem, collecting/cleaning data, and monitoring after deployment are also important. The three stages above are the assignment's requested focus.

## 5. Flowchart: Machine Learning Workflow

Run the following cell in Google Colab to display a flowchart. The three stages requested by the assignment are highlighted within the full workflow.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

steps = [
    ("Define problem & collect data", "#e5e7eb"),
    ("Prepare data", "#e5e7eb"),
    ("1. Feature Selection", "#bfdbfe"),
    ("2. Model Selection & Training", "#bfdbfe"),
    ("3. Model Evaluation", "#bfdbfe"),
    ("Improve / Deploy & Monitor", "#e5e7eb"),
]
fig, ax = plt.subplots(figsize=(10, 9))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis("off")
for i, (label, color) in enumerate(steps):
    y = 8.6 - i * 1.5
    box = FancyBboxPatch((2.0, y), 6.0, 0.85,
                         boxstyle="round,pad=0.1,rounding_size=0.15",
                         linewidth=1.5, edgecolor="#334155", facecolor=color)
    ax.add_patch(box)
    ax.text(5.0, y + 0.425, label, ha="center", va="center", fontsize=12, weight="bold")
    if i < len(steps) - 1:
        ax.add_patch(FancyArrowPatch((5.0, y - 0.05), (5.0, y - 0.52),
                                    arrowstyle="-|>", mutation_scale=18,
                                    linewidth=1.7, color="#475569"))
ax.set_title("Machine Learning Development Process", fontsize=17, weight="bold", pad=18)
plt.tight_layout()
plt.show()
```

### Interpreting the flowchart

The process starts by defining the objective and preparing suitable data. **Feature selection** identifies relevant predictors; **model selection and training** choose and fit an algorithm; **model evaluation** checks performance on unseen examples. Weak evaluation results should lead to revisiting features, model settings, or data quality. Once deployed, performance should continue to be monitored.

## 6. Conclusion

Machine learning helps analysts move beyond reporting past events toward predicting outcomes, discovering patterns, and supporting better decisions. **Supervised learning** uses labeled examples, **unsupervised learning** finds structure in unlabeled data, and **reinforcement learning** learns action strategies through rewards.

Successful projects require thoughtful **feature selection**, a model suited to the problem, and rigorous **evaluation**. No algorithm is automatically best: the right approach depends on the business goal, data quality, available labels, and consequences of incorrect predictions.

### Reflection

A key lesson is that choosing a sophisticated algorithm is less important than correctly defining the problem and evaluating the model on unseen data. For example, high classification accuracy may be misleading if the target class is rare. Careful feature selection, appropriate metrics, and ongoing monitoring make ML results more useful and trustworthy.
